```python
import cv2
import os

def extract_frames(video_path, output_folder, frame_interval=1):
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"Error: Could not open video {video_path}")
        return

    os.makedirs(output_folder, exist_ok=True)
    frame_count = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        if frame_count % frame_interval == 0:
            frame_name = os.path.join(output_folder, f"frame_{frame_count:05d}.jpg")
            cv2.imwrite(frame_name, frame)
        frame_count += 1
    cap.release()
    print(f"Extracted {frame_count // frame_interval} frames from {video_path}")

# 예시 사용법: YawDD 데이터셋에서 비디오 프레임 추출
# YawDD 데이터셋 경로: /root/.cache/kagglehub/datasets/enider/yawdd-dataset/versions/4
# 'ls -R' 명령어로 확인한 실제 비디오 파일들이 있는 경로들을 지정합니다.
yawdd_root_path = "/root/.cache/kagglehub/datasets/enider/yawdd-dataset/versions/4"
video_folders = [
    os.path.join(yawdd_root_path, "Dash/Dash/Female"),
    os.path.join(yawdd_root_path, "Dash/Dash/Male"),
    os.path.join(yawdd_root_path, "Mirror/Mirror/Female_mirror"),
    # 'Male_mirror Avi Videos' 폴더 이름에 공백이 있으므로 주의
    os.path.join(yawdd_root_path, "Mirror/Mirror/Male_mirror Avi Videos")
]
output_frames_dir = "yawdd_extracted_frames"
os.makedirs(output_frames_dir, exist_ok=True)

# YawDD 데이터셋의 모든 비디오 파일에 대해 프레임 추출을 반복합니다.
for folder_path in video_folders:
    if not os.path.exists(folder_path):
        print(f"Warning: Video folder not found: {folder_path}. Skipping.")
        continue

    for video_file_name in os.listdir(folder_path):
        # 파일 이름에 공백이 있는 경우를 위해 strip() 사용
        clean_video_file_name = video_file_name.strip()
        if clean_video_file_name.endswith(('.avi', '.mp4')): # 비디오 파일 확장자 확인
            video_path = os.path.join(folder_path, clean_video_file_name)

            # 각 비디오 파일의 이름을 따서 프레임을 저장할 하위 폴더 생성
            # 폴더 이름에 특수 문자나 공백이 포함될 수 있으므로, 안전한 이름으로 변환
            safe_folder_name = os.path.splitext(clean_video_file_name)[0].replace(" ", "_").replace("-", "_")
            current_video_output_folder = os.path.join(output_frames_dir, safe_folder_name)

            extract_frames(video_path, current_video_output_folder, frame_interval=5) # 5프레임마다 추출
```

```python
import cv2
import dlib
import numpy as np
import os

# dlib 얼굴 검출기 및 랜드마크 예측기 초기화
detector = dlib.get_frontal_face_detector()

predictor_path = "shape_predictor_68_face_landmarks.dat"
if not os.path.exists(predictor_path):
    print(f"Error: {predictor_path} not found. Please download it using the commands above.")
    predictor = None # 예측기 로드 실패 시 None으로 설정
else:
    try:
        predictor = dlib.shape_predictor(predictor_path)
    except Exception as e:
        print(f"Error loading predictor: {e}. Make sure '{predictor_path}' is correctly placed.")
        predictor = None # 예측기 로드 실패 시 None으로 설정

def get_landmarks(image):
    if predictor is None:
        return None, None # 예측기가 로드되지 않았으면 처리하지 않음

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    faces = detector(gray)
    if len(faces) == 0:
        return None, None
    # 가장 큰 얼굴 선택
    face = max(faces, key=lambda rect: rect.area())
    landmarks = predictor(gray, face)
    return face, landmarks

def extract_eye_roi(image, landmarks, eye_points_start, eye_points_end):
    # 눈 랜드마크 좌표 추출
    points = []
    for n in range(eye_points_start, eye_points_end + 1):
        x = landmarks.part(n).x
        y = landmarks.part(n).y
        points.append((x, y))

    # 눈 영역의 바운딩 박스 계산
    points = np.array(points)
    x_min, y_min = np.min(points, axis=0)
    x_max, y_max = np.max(points, axis=0)

    # 여유 공간 추가 (선택 사항)
    padding = 10
    x_min = max(0, x_min - padding)
    y_min = max(0, y_min - padding)
    x_max = min(image.shape[1], x_max + padding)
    y_max = min(image.shape[0], image.shape[0] - 1 if image.shape[0] > 0 else 0) # y_max가 이미지 높이를 초과하지 않도록 보정

    # 유효한 ROI 범위 확인
    if x_max <= x_min or y_max <= y_min:
        return None # 유효하지 않은 ROI 반환

    return image[y_min:y_max, x_min:x_max]

def extract_mouth_roi(image, landmarks):
    # 입 랜드마크 좌표 (dlib 랜드마크 인덱스 48-67)
    points = []
    for n in range(48, 68):
        x = landmarks.part(n).x
        y = landmarks.part(n).y
        points.append((x, y))

    points = np.array(points)
    x_min, y_min = np.min(points, axis=0)
    x_max, y_max = np.max(points, axis=0)

    padding = 10
    x_min = max(0, x_min - padding)
    y_min = max(0, y_min - padding)
    x_max = min(image.shape[1], x_max + padding)
    y_max = min(image.shape[0], image.shape[0] - 1 if image.shape[0] > 0 else 0) # y_max가 이미지 높이를 초과하지 않도록 보정

    # 유효한 ROI 범위 확인
    if x_max <= x_min or y_max <= y_min:
        return None # 유효하지 않은 ROI 반환

    return image[y_min:y_max, x_min:x_max]
```

```python
def resize_image(image, size=(64, 64)): #이미지 크기 조정, 64*64사이즈
    return cv2.resize(image, size)
```

```python
def normalize_image(image): #정규화 진행, 학습 효율 높이기 위함. -1~1
    return image.astype("float32") / 255.0
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator # 데이터의 양을 늘리고 모델의 일반화 성능을 향상시킴

# 이미지 데이터 증강 설정
datagen = ImageDataGenerator(
    rotation_range=20,        # 이미지 회전 (각도)
    width_shift_range=0.2,    # 가로 이동 (비율)
    height_shift_range=0.2,   # 세로 이동 (비율)
    shear_range=0.2,          # 전단 변환
    zoom_range=0.2,           # 확대/축소
    horizontal_flip=True,     # 좌우 반전
    fill_mode='nearest'       # 빈 공간 채우기 방식
)
```
